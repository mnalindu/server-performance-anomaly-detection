```python
# Import required libraries

import numpy as np
import pandas as pd
from pathlib import Path

print("Libraries imported successfully!")
```

```python
# Locate the dataset folder

possible_paths = [
    Path("data/raw"),
    Path("../data/raw")
]

DATA_DIR = next(
    (path for path in possible_paths
     if (path / "train.csv").exists()),
    None
)

if DATA_DIR is None:
    raise FileNotFoundError(
        "Dataset files not found. Check data/raw folder."
    )

print("Dataset location:", DATA_DIR.resolve())


# Load CSV files

train_df = pd.read_csv(DATA_DIR / "train.csv")
test_df = pd.read_csv(DATA_DIR / "test.csv")
test_label_df = pd.read_csv(DATA_DIR / "test_label.csv")

print("\nAll PSM datasets loaded successfully!")
```

```python
# Analyze missing values in each training feature

missing_summary = pd.DataFrame({
    "Missing Count": train_df.isnull().sum(),
    "Missing Percentage": (
        train_df.isnull().sum() / len(train_df)
    ) * 100
})

# Display only columns containing missing values

missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
].sort_values(
    by="Missing Count",
    ascending=False
)

display(missing_summary)
```

```python
# Analyze missing value patterns

feature_cols = [
    col for col in train_df.columns
    if col.startswith("feature_")
]

# Count affected rows
affected_rows = train_df[feature_cols].isnull().any(axis=1).sum()

print("Total training rows:", len(train_df))
print("Rows containing missing values:", affected_rows)

print(
    "Percentage of affected rows:",
    round(affected_rows / len(train_df) * 100, 3),
    "%"
)

# Analyze consecutive missing values
print("\nCONSECUTIVE MISSING VALUE ANALYSIS")

for col in missing_summary.index:

    mask = train_df[col].isnull()

    groups = mask.ne(mask.shift()).cumsum()

    run_lengths = mask[mask].groupby(
        groups[mask]
    ).size()

    print(f"\n{col}")
    print("Number of missing groups:", len(run_lengths))
    print("Longest consecutive gap:", run_lengths.max())
```

```python
# Identify exact locations of consecutive missing values

missing_blocks = []

for col in feature_cols:

    positions = np.flatnonzero(
        train_df[col].isna().to_numpy()
    )

    if len(positions) == 0:
        continue

    # Separate consecutive missing positions
    blocks = np.split(
        positions,
        np.where(np.diff(positions) > 1)[0] + 1
    )

    for block in blocks:

        start_pos = int(block[0])
        end_pos = int(block[-1])

        missing_blocks.append({
            "Feature": col,
            "Start Row": start_pos,
            "End Row": end_pos,
            "Start Timestamp": train_df.iloc[start_pos]["timestamp_(min)"],
            "End Timestamp": train_df.iloc[end_pos]["timestamp_(min)"],
            "Missing Count": len(block)
        })

# Create summary table
blocks_df = pd.DataFrame(missing_blocks)

blocks_df = blocks_df.sort_values(
    ["Feature", "Start Row"]
).reset_index(drop=True)

display(blocks_df)
```

```python
# Count missing features per row

missing_per_row = train_df[feature_cols].isnull().sum(axis=1)

print("Missing features per row:")
print(missing_per_row.value_counts().sort_index())

print("\nRows with multiple missing features:")

print(
    (missing_per_row > 1).sum()
)
```

```python
# Inspect values immediately before and after missing gaps

intervals = [
    ("feature_21", 3961, 4319),
    ("feature_21", 18001, 19573),
    ("feature_21", 20881, 21084),
    ("feature_6", 71641, 71999),
    ("feature_13", 85625, 85826)
]

for feature, start, end in intervals:

    print(f"\n{'='*60}")
    print(f"Feature: {feature}")
    print(f"Missing interval: {start} - {end}")

    columns = ["timestamp_(min)", feature]

    before = train_df.iloc[
        max(0, start-2):start
    ][columns].copy()

    after = train_df.iloc[
        end+1:end+3
    ][columns].copy()

    before.insert(0, "Position", before.index)
    after.insert(0, "Position", after.index)

    print("\nValues BEFORE missing interval:")
    display(before)

    print("Values AFTER missing interval:")
    display(after)
```

```python
import matplotlib.pyplot as plt

# Important missing intervals
intervals = [
    ("feature_21", 3961, 4319),
    ("feature_21", 18001, 19573),
    ("feature_21", 20881, 21084),
    ("feature_6", 71641, 71999),
    ("feature_13", 85625, 85826)
]

for feature, start, end in intervals:

    # Get 100 observations before the gap
    before = train_df.iloc[
        max(0, start-100):start
    ]

    # Get 100 observations after the gap
    after = train_df.iloc[
        end+1:min(len(train_df), end+101)
    ]

    fig, axes = plt.subplots(1, 2, figsize=(13, 3))

    # Before missing interval
    axes[0].plot(
        before["timestamp_(min)"],
        before[feature]
    )

    axes[0].set_title("Before Missing Interval")
    axes[0].set_xlabel("Timestamp")
    axes[0].set_ylabel(feature)
    axes[0].grid(alpha=0.3)

    # After missing interval
    axes[1].plot(
        after["timestamp_(min)"],
        after[feature]
    )

    axes[1].set_title("After Missing Interval")
    axes[1].set_xlabel("Timestamp")
    axes[1].set_ylabel(feature)
    axes[1].grid(alpha=0.3)

    fig.suptitle(
        f"{feature}: Missing interval {start} - {end}"
    )

    plt.tight_layout()
    plt.show()
```
